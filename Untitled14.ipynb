```python
valorCompra = float(input('Digite o valor da compra '))
print(f'O valor da compra é {valorCompra} ')

cartaoLoja = input('Possui cartão da loja? Digite: sim / não ')
if cartaoLoja == 'sim':
  print('Possui cartão. ')
elif cartaoLoja == 'não':
  print('Não possui cartão. ')

desconto = valorCompra * 0.10

if valorCompra >= 100 and cartaoLoja == 'sim':
  print(f'Você recebeu um desconto de 10% na sua compra. {desconto}')
  print(f'O valor com desconto é: {valorCompra - desconto}' )
else:
  print('Você não recebeu desconto. ')
```

```python
valor = float(input("Digite o valor da compra"))
cartao = input("Possui cartão sim ou não?")

if valor >= 100 and cartao == "sim":
  desconto = valor * 0.10
  valor_total = valor - desconto
  print(f"O desconto será {desconto}")
  print(f"O valor final será {valor_total}")
else:
  print("Desconto não aplicado")
  print(f"O valor final {valor}")
```
